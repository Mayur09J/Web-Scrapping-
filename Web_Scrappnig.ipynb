```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
url='https://www.screener.in/screens/717439/the-top-500-indian-companies/?page=1'
headers={'connection':'stay-alive'}
webpage=requests.get(url=url,headers=headers)
```

```python
soup=BeautifulSoup(webpage.content,'lxml')
```

```python
table=soup.find('table')
```

```python
columns=[colu.get_text(strip=True) for colu in table.find_all('th')]
```

```python
Company_info=pd.DataFrame(columns=columns[:11])
```

```python

# Create a single master list to hold rows from ALL pages
all_companies = []

for p in range(1, 52):
    url = 'https://www.screener.in/screens/717439/the-top-500-indian-companies/?page={}'.format(p)
    headers = {'User-Agent': 'Mozilla/5.0'} # Created the header to avoid getting blocked

    webpage = requests.get(url=url, headers=headers)
    soup = BeautifulSoup(webpage.content, 'lxml')

    table = soup.find('table')

    # Skip to next page if table is not found (e.g., if a page doesn't exist)
    if not table:
        continue

    for i in table.find_all('tr'):
        # get_text(strip=True) automatically cleans up '\n' and spaces!
        columns = [col.get_text(strip=True) for col in i.find_all('td')]

        if columns:
            # Append directly to the master list
            all_companies.append(columns)

# 2. Convert the entire master list to a DataFrame AT THE VERY END (Outside all loops)
company_info = pd.DataFrame(all_companies)



```

```python
company_info.set_index(0,inplace=True)
```

```python
# Updated column names
company_info.rename(columns={0:'Sr.No',1:'Company_name',2:'Cmp Rs',3:'P/E',4:'Mar Cap Rr Cr',5:'Div Yid %',6:'NP Qtr Rs Cr',7:'QTR PROFIT Var%',8:'Sles Qtr Rs.Cr',9:'Qtr Sales Var%',10:'ROCE%'},inplace=True)
```
