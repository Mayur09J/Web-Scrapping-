```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random

all_bowlers = []
headers_list = []
max_pages = 42

i = 1
while i <= max_pages:
   url = r"https://stats.cricinfo.com/ci/engine/stats/index.html?class=11;filter=advanced;home_or_away=1;home_or_away=2;home_or_away=3;orderby=wickets;page={};result=1;result=2;result=3;result=4;result=5;size=200;template=results;type=bowling".format(i)
   headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36',
        'Referer': 'https://google.com',
        'Connection': 'keep-alive'
    }

    # Attempt to fetch the page with up to 3 retries if it hits a wall
   for attempt in range(3):
        try:
            response = requests.get(url=url, headers=headers, timeout=10)
            soup = BeautifulSoup(response.content, 'html.parser')

            main_table = soup.find(lambda tag: tag.name == 'table'
                                   and 'engineTable' in tag.get('class', [])
                                   and not tag.has_attr('style'))

            if main_table:
                # 1. Grab columns if we don't have them yet
                if not headers_list:
                    header_row = main_table.find('tr', class_='headLinks')
                    if header_row:
                        headers_list = [th.get_text(strip=True) for th in header_row.find_all('th')]

                # 2. Extract data rows
                rows = main_table.find_all('tr', class_='data1')

                if rows:
                    for row in rows:
                        data = [td.get_text(strip=True) for td in row.find_all('td')]
                        if data:
                            all_bowlers.append(data)

                    print(f"✅ Successfully scraped page {i}/{max_pages} (Total rows collected: {len(all_bowlers)})")
                    i += 1  # Move to the next page only on success
                    break

            # If the table or rows are missing, we might be getting soft-blocked
            print(f"⚠️ Page {i} layout missing table/rows. Retrying attempt {attempt + 1}...")
            time.sleep(random.uniform(3, 6)) # Pause longer before retrying

        except Exception as e:
              print(f"❌ Connection error on page {i}: {e}. Retrying...")
              time.sleep(5)
        else:
          print(f"🚨 Failed to scrape page {i} after 3 attempts. Stopping to preserve collected data.")
          break

      # Randomizing sleep delay between 1.5 to 3 seconds keeps the script under the radar
        time.sleep(random.uniform(1.5, 3.0))

  # Convert to DataFrame
if all_bowlers and headers_list:
  df = pd.DataFrame(all_bowlers, columns=headers_list)
  df.to_csv("cricinfo_bowling_stats_complete.csv", index=False)
  print(f"🎉 Process Complete! Successfully saved {len(df)} rows to 'cricinfo_bowling_stats_complete.csv'")
else:
  print("Failed to save data. No records collected.")
```
